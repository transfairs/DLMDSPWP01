# Import libraries

```python
import FileReader as f
import Regression as r
from SQL import PersistenceUtil
import statsmodels.api as sm
from statsmodels.nonparametric.kernel_regression import KernelReg
import Visualisation as visual

import numpy as np
```

```python
class FilterVisualizer(visual.Visualizer):
    def additional_elements(self, index, min_y):
        pass

def convolution(y, length, mode="same"):
    y2 = np.ones(length)/length
    #zeros = np.zeros(len(y)-length)
    #y2 = np.concatenate((ones, zeros), axis=0)
    return np.convolve(y, y2, mode=mode)

def regression(y):
    kr = KernelReg(y,y.index,'c')
    y, y_std = kr.fit(y.index)

    return y

def hodrick_prescott(y):
    cycle, trend = sm.tsa.filters.hpfilter(y, 1600)

    return cycle, trend

def wrangling(df_train, df_ideal, title="Training Data", plot=False):
    df_train.index.names = ['x']
    df_ideal.index.names = ['x']
    regression = r.Regression(df_ideal)
    df_train_mse = df_train.apply(regression.mse)
    df_result = regression.min_value(df_train_mse, df_train)
    chosen_ideal = df_ideal[df_result.min_y.tolist()]

    if plot:
        visual.PlotFactory.show(title, df_train, df_result, chosen_ideal)
    
    return df_result

def main():
    pu = PersistenceUtil()
    df_train = pu.read_df("train").set_index("X");
    df_ideal = pu.read_df("ideal").set_index("X");

    #print(df_ideal)
    df_train_convolve = df_train.copy()
    support_len = 12
    df_train_convolve = df_train_convolve.apply(lambda y: convolution(y, support_len))

    i = int(support_len/2)

    df_train_convolve = df_train_convolve.iloc[i:-i]

    df_result = wrangling(df_train, df_ideal)
    df_ideal_convolve = df_ideal.iloc[i:-i]
    df_result_convolve = wrangling(df_train_convolve, df_ideal_convolve, plot=False)

#    df_train_regression = df_train.copy()
#    df_train_regression = df_train_regression.apply(regression)
#    df_result_regression = wrangling(df_train_regression, df_ideal, plot=False)

    df_train_hp = df_train.copy()
    df_ideal_hp = df_ideal[df_result.min_y]
#    df_ideal_hp = df_ideal.copy()
    df_train_hp_trend = df_train_hp.apply(lambda y: hodrick_prescott(y)[0])
    df_train_hp_cycle = df_train_hp.apply(lambda y: hodrick_prescott(y)[1])
    
    df_train_hp = df_train_hp-df_train_hp_trend-df_train_hp_cycle
    df_train_hp_trend.columns = df_result.min_y
    df_train_hp_cycle.columns = df_result.min_y
    print(df_ideal_hp)
    df_ideal_hp = df_ideal_hp-df_train_hp_trend-df_train_hp_cycle
    print("\ndf_ideal_hp\n", df_ideal_hp)
#    df_ideal_hp = df_ideal_hp.fillna(0)

    df_result_hp = wrangling(df_train_hp, df_ideal_hp, plot=False)

    print(df_result)
#    print(df_ideal)
#    print(df_result_convolve)
#    print(df_result_regression)
    print("\ndf_result_hp\n", df_result_hp)

#    print("\ndf_ideal_hp\n", df_ideal_hp)
#    print("\ndf_train_hp_trend\n", df_train_hp_trend)
#    print("\ndf_train_hp\n", df_train_hp)
#    print("\ndf_train_hp_trend\n", df_ideal_hp-df_train_hp_trend)
    print("Done")

if __name__ == '__main__':
    main()
```
