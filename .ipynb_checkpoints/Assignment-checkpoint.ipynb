# Setup

```python
# Set log level to DEBUG while developing
log_level = logging.DEBUG

def app_init():
    '''
    Initialises the logger.
    '''
    logging.basicConfig(filename="app.log", filemode="a", \
                        format="[%(asctime)s] %(name)-6s - %(levelname)-8s - %(message)s", datefmt='%Y-%m-%d %H:%M:%S')

    logger = logging.getLogger()
    logger.setLevel(log_level)

    logger.info("App started")

    # Test the logger in DEBUG and NOTSET mode.
    if logger.level < logging.INFO:
        logger.debug("Debugging")
        logger.warning("Test logger")

app_init()
```

# Read CSV

```python
def read_csv_to_pd(file=None):
    output = None
    try:
        output = pd.read_csv(filepath_or_buffer=file)
        output.set_index('x', inplace=True)
        message = "File loaded: {file}"
    except FileNotFoundError as e:
        message = "File not found: {file}"
    except:
        message = "An unknown error occured with file: {file}."
    finally:
        logger.info(message.format(file=file))
    return output




if __name__ == '__main__':
    pd_train = read_csv_to_pd("data/train.csv")
    pd_ideal = read_csv_to_pd("data/ideal.csv")
```

```python
def mse(training_data):
    output = pd_ideal.apply(mean_squared_error,y_pred=training_data)
    return output

def min_mse(df):
    output = df.apply(min).to_frame("min_value")
    output = output.assign(min_y = df.idxmin())
    return output

df = pd_train.apply(mse)

print(min_mse(df))
```

# SQL

```python
# Get engine object using pymysql driver for mysql
engine = db.create_engine(f"sqlite:///data/sqlite.db")

# Get connection object
connection = engine.connect()

# Get MetaData object
meta_data = db.MetaData()

# Set training data creation script table
train_table = db.Table("train", meta_data, \
                       db.Column("X", db.String(5), primary_key=True, autoincrement=False, nullable=False), \
                       db.Column("Y1 (training func)", db.Integer, nullable=False), \
                       db.Column("Y2 (training func)", db.Integer, nullable=False), \
                       db.Column("Y3 (training func)", db.Integer, nullable=False), \
                       db.Column("Y4 (training func)", db.Integer, nullable=False), \
                      )

# create actor table and stores the information in metadata
meta_data.create_all(engine)

connection.close()
```

```python
def sql_column(y_string, replacement):
    return re.sub("y(\d)",replacement,y_string)

engine = db.create_engine(f"sqlite:///data/sqlite.db")

def save_df_to_db(df, db, table, regex_column_name):
    df_sql = df.copy()
    df_sql.columns = df_sql.columns.map(lambda c: sql_column(c, regex_column_name))
    df_sql.index.names = ['X']

    return df_sql.to_sql(table, db, if_exists="replace", index=True)

save_df_to_db(pd_train, engine, "train", r"Y\1 (training func)")
save_df_to_db(pd_ideal, engine, "ideal", r"Y\1 (ideal func)")
```
